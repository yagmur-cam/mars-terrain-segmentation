# 03 – Model tests: U-Net shapes and metrics

Checks for `src/unet.py` and `src/metrics.py`: output shapes of `DoubleConv` and the full `UNet` on a fake batch, and the confusion matrix / IoU functions on a small hand-made example.

**Results:** `UNet` maps `[2, 1, 512, 512]` → `[2, 4, 512, 512]` with 31,036,676 parameters. The hand-made example gives the expected IoU `[0.5, 0.5, nan, nan]` and mIoU 0.5.

```python
import sys
sys.path.append("..")   # the repo root, where src/ lives

import torch 
from src.unet import DoubleConv

# testing DoubleConv
block = DoubleConv(1, 64)
x = torch.randn(2, 1, 512, 512)   # fake batch: 2 greyscale images
print(block(x).shape) # # expect torch.Size([2, 64, 512, 512])
```

```python
from src.unet import UNet 

# testing UNet 
model = UNet()
print(model(torch.randn(2, 1, 512, 512)).shape)          # expect [2, 4, 512, 512]
print(sum(p.numel() for p in model.parameters()))        # number of parameters, 31 million 
```
